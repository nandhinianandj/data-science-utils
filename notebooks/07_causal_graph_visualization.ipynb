# Causal Graph Visualization Demo

This notebook demonstrates the new **CausalGraph** capabilities:
- Building causal graphs programmatically
- Domain knowledge injection (freezing edges)
- Graph visualization (DOT export)
- Parent/child queries

```rust
:dep datascienceutils-core = { path = "../datascienceutils-core", features = ["causal-analysis"] }
```

```rust
use datascienceutils_core::analyze::CausalGraph;
```

## Example 1: Medical Treatment Study

Build a causal graph for a medical study:
- **Age** → **Treatment** (older patients more likely to receive treatment)
- **Age** → **Outcome** (age affects health outcomes)
- **Treatment** → **Outcome** (treatment affects health)

```rust
let mut graph = CausalGraph::new();

// Add edges
graph.add_edge("Age", "Treatment").unwrap();
graph.add_edge("Age", "Outcome").unwrap();
graph.add_edge("Treatment", "Outcome").unwrap();

println!("Medical Study Graph:");
println!("Nodes: {}", graph.num_nodes());
println!("Edges: {}", graph.num_edges());
println!("\nParents of Outcome: {:?}", graph.parents("Outcome"));
println!("Children of Age: {:?}", graph.children("Age"));
```

## Example 2: E-commerce Conversion

Model factors affecting online conversion:
- **UserAge** → **AdClick**
- **UserAge** → **Purchase**
- **AdClick** → **Purchase**
- **Price** → **Purchase**

```rust
let mut ecommerce_graph = CausalGraph::new();

ecommerce_graph.add_edge("UserAge", "AdClick").unwrap();
ecommerce_graph.add_edge("UserAge", "Purchase").unwrap();
ecommerce_graph.add_edge("AdClick", "Purchase").unwrap();
ecommerce_graph.add_edge("Price", "Purchase").unwrap();

println!("E-commerce Graph:");
println!("Nodes: {}", ecommerce_graph.num_nodes());
println!("Edges: {}", ecommerce_graph.num_edges());
println!("\nFactors affecting Purchase: {:?}", ecommerce_graph.parents("Purchase"));
```

## Example 3: Domain Knowledge Injection

Freeze edges based on expert knowledge (e.g., known causal relationships).

```rust
let mut expert_graph = CausalGraph::new();

// Build graph
expert_graph.add_edge("Smoking", "LungCancer").unwrap();
expert_graph.add_edge("Genetics", "LungCancer").unwrap();
expert_graph.add_edge("Age", "LungCancer").unwrap();

// Freeze the well-established Smoking → LungCancer edge
expert_graph.freeze_edge("Smoking", "LungCancer").unwrap();

println!("Expert Knowledge Graph:");
println!("Frozen edge: Smoking → LungCancer");
println!("\nAttempting to modify frozen edge...");

match expert_graph.add_edge_weighted("Smoking", "LungCancer", 2.0) {
    Ok(_) => println!("✗ Edge modified (unexpected)"),
    Err(e) => println!("✓ Edge protected: {}", e),
}
```

## Example 4: DOT Export for Visualization

Export graphs to Graphviz DOT format for rendering.

```rust
let mut viz_graph = CausalGraph::new();

// Build a complex graph
viz_graph.add_edge("Education", "Income").unwrap();
viz_graph.add_edge("Education", "Health").unwrap();
viz_graph.add_edge("Income", "Health").unwrap();
viz_graph.add_edge("Income", "LifeSatisfaction").unwrap();
viz_graph.add_edge("Health", "LifeSatisfaction").unwrap();

// Freeze known relationships
viz_graph.freeze_edge("Education", "Income").unwrap();

println!("DOT Format Output:");
println!("{}", viz_graph.to_dot());
println!("\n✓ Save this to a .dot file and render with: dot -Tpng graph.dot -o graph.png");
```

## Example 5: Weighted Edges

Add edge weights to represent strength of causal relationships.

```rust
let mut weighted_graph = CausalGraph::new();

// Strong causal effect
weighted_graph.add_edge_weighted("Exercise", "Fitness", 0.8).unwrap();

// Moderate effect
weighted_graph.add_edge_weighted("Diet", "Fitness", 0.5).unwrap();

// Weak effect
weighted_graph.add_edge_weighted("Sleep", "Fitness", 0.3).unwrap();

println!("Weighted Causal Graph:");
println!("{}", weighted_graph.to_dot());
```
